```python
import os
import glob
import pandas as pd

# 1. Get a list of all CSV file paths in the '../data/' folder.
file_paths = glob.glob('../data/*.csv')

# 2. Create an empty dictionary to store the loaded DataFrames.
datasets = {}

for path in file_paths:
    # Extract just the file name from the path.
    # (e.g., '../data/olist_customers_dataset.csv' -> 'olist_customers_dataset')
    file_name = os.path.basename(path).replace('.csv', '')
    
    # Read the CSV file into a DataFrame.
    df = pd.read_csv(path)
    
    # Save it to the dictionary. (datasets['file name'] = DataFrame)
    datasets[file_name] = df
    
    # 3. Print a quick summary of the data.
    print(f"📑 File Name: {file_name}")
    print(f" - Data Shape (Rows, Columns): {df.shape}")
    print(f" - Column List: {df.columns.tolist()}")
    print("-" * 50)
```

```python
from IPython.display import display, Markdown
import io

selected_names = [
    "olist_orders_dataset",
    "olist_order_items_dataset",
    "olist_customers_dataset",
    "olist_products_dataset",
    "product_category_name_translation",
    "olist_order_reviews_dataset",
]

for name in selected_names:
    df = datasets[name]

    display(Markdown(f"# 📑 Dataset: `{name}`"))

    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())


    display(Markdown("## Head"))
    display(df.head())

    display(Markdown("## Tail"))
    display(df.tail())

    display(Markdown("## Info"))
    df.info()

    display(Markdown("## Describe"))
    display(df.describe().T)

    print("=" * 100)
```

```python
# 1. Get the datasets we need
orders = datasets["olist_orders_dataset"]
items = datasets["olist_order_items_dataset"]
products = datasets["olist_products_dataset"]

# 2. Filter only 'delivered' orders
delivered_orders = orders[orders['order_status'] == 'delivered']

# 3. Merge datasets: Orders + Items + Products
# a. Merge delivered orders and items
merged_df = pd.merge(delivered_orders, items, on='order_id', how='inner')

# b. Merge the result with products
merged_df = pd.merge(merged_df, products, on='product_id', how='inner')

# 4. Find rows where the category name is missing (NaN)
# We can use .isnull() to find empty values
missing_category_df = merged_df[merged_df['product_category_name'].isnull()]

# Check the total number of missing rows
print(f"Total missing rows: {len(missing_category_df)}\n")

# Select only the columns we want to see
cols_to_show = ['order_id', 'product_id', 'order_status', 'product_category_name']
display(missing_category_df[cols_to_show].head(10))
display(missing_category_df[cols_to_show].tail(10))
```

```python
from IPython.display import Markdown

# 1. Get the total number of rows in your merged data
total_count = len(merged_df)

# 2. Get the number of missing (NULL) category names
null_count = merged_df['product_category_name'].isnull().sum()

# 3. Calculate the NULL rate formula
null_rate = (null_count / total_count) * 100

# 4. Display the result nicely in Jupyter Notebook
display(Markdown("### 📊 NULL Rate Report (product_category_name)"))
print(f"1. Total Data Count : {total_count:,} rows")
print(f"2. NULL Data Count  : {null_count:,} rows")
print("-" * 40)
print(f"🔥 Final NULL Rate  : {null_rate:.2f}%")
```
